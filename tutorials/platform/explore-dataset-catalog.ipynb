## Explore the dataset catalogue

`list_datasets()` returns all the datasets in the Cecil catalog. Use it to browse what is available, filter by topic, or validate that a dataset will accept your AOI before creating a subscription.

```python
import pandas as pd
import cecil

client = cecil.Client()
```

### Browse the catalogue

Build a summary table to get an overview of available datasets.

```python
datasets = client.list_datasets()
```

```python
pd.DataFrame([
    {
        'name': d.name,
        'licence': d.licence.type,
        'type': d.type,
        'provider': ', '.join(p.name for p in d.providers),
        'categories': ', '.join(d.categories),
        'spatial_resolution': d.spatial_resolution.nominal,
        'temporal_coverage': d.temporal_coverage.nominal,
    }
    for d in datasets
])
```

### Filter by category

Each dataset has a `categories` list. Use it to find datasets relevant to a particular topic. Categories may contain multi-word strings, so check for substring matches.

```python
risk_datasets = [d for d in datasets if 'Risk' in d.categories]

for d in risk_datasets:
    print(d.name)
    print(f"  {d.description[0]}")
```

### Check for variables of interest
Filter the catalog to check for specific variables of interest across datasets.

```python
forest_cover_datasets = [d for d in datasets if any(v.name == 'forest_cover' for v in d.variables)]

for d in forest_cover_datasets:
    print(d.name)
```

### Check AOI compatibility

Dataset constraints define the AOI sizes they accept. Filter the catalogue to find datasets that are compatible with your AOI area before creating a subscription.

```python
my_aoi_hectares = 50_000

compatible = [
    d for d in datasets
    if (d.constraints.aoi_min_hectares is None or d.constraints.aoi_min_hectares <= my_aoi_hectares)
    and (d.constraints.aoi_max_hectares is None or d.constraints.aoi_max_hectares >= my_aoi_hectares)
]

print(f"{len(compatible)} of {len(datasets)} datasets accept an AOI of {my_aoi_hectares:,} ha:\n")
for d in compatible:
    print(f"  {d.name}")
```
